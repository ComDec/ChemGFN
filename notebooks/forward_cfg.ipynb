```python
from transformers import AutoTokenizer
from transformers_cfg.recognizer import StringRecognizer
from transformers_cfg.parser import parse_ebnf
```

```python
model_id = "meta-llama/Llama-3.2-1B"
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
with open(f"/data1/xw3763/project/gflow/ChemGFN/assets/24_grammars/var_length.ebnf", "r") as file:
    grammar_str = file.read()

parsed_grammar = parse_ebnf(grammar_str)
```

```python
string_grammar = StringRecognizer(
        parsed_grammar.grammar_encoding, parsed_grammar.symbol_table["root"]
    )
```

```python
from chemgfn.models.validators import Expr24Validator
validator = Expr24Validator(scorer="hit24")
validator_near = Expr24Validator(scorer="near_24")
validator_dense = Expr24Validator(scorer="hit24_dense")
validator_near_dense = Expr24Validator(scorer="near_24_dense")
```

```python
sample_1 = "1+1*2*1"
sample_1_tokens = tokenizer.encode(sample_1, add_special_tokens=False, return_tensors="pt")
sample_2 = "3*5"
sample_2_tokens = tokenizer.encode(sample_2, add_special_tokens=False, return_tensors="pt")
sample_3 = "8*3+1"
sample_3_tokens = tokenizer.encode(sample_3, add_special_tokens=False, return_tensors="pt")
sample_4 = "6+6+6+5"
sample_4_tokens = tokenizer.encode(sample_4, add_special_tokens=False, return_tensors="pt")
```

```python
print("hit24", validator(sample_1_tokens, tokenizer))
print("near24", validator_near(sample_1_tokens, tokenizer))
print("hit24_dense", validator_dense(sample_1_tokens, tokenizer))
print("near24_dense", validator_near_dense(sample_1_tokens, tokenizer))
```

```python
print("hit24", validator(sample_2_tokens, tokenizer))
print("near24", validator_near(sample_2_tokens, tokenizer))
print("hit24_dense", validator_dense(sample_2_tokens, tokenizer))
print("near24_dense", validator_near_dense(sample_2_tokens, tokenizer))

print("hit24", validator(sample_3_tokens, tokenizer))
print("near24", validator_near(sample_3_tokens, tokenizer))
print("hit24_dense", validator_dense(sample_3_tokens, tokenizer))
print("near24_dense", validator_near_dense(sample_3_tokens, tokenizer))

print("hit24", validator(sample_4_tokens, tokenizer))
print("near24", validator_near(sample_4_tokens, tokenizer))
print("hit24_dense", validator_dense(sample_4_tokens, tokenizer))
print("near24_dense", validator_near_dense(sample_4_tokens, tokenizer))
```
