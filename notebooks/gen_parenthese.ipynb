```python
import random

def generate_brackets(max_depth=3):
    """
    生成合法且复杂的括号组合字符串
    :param max_depth: 控制嵌套深度的参数
    :return: 生成的括号字符串
    """
    if max_depth <= 0:
        # 当达到最大深度时返回基本括号对
        return random.choice(['()', '[]', '<>'])
    
    # 随机选择生成模式：0-嵌套结构，1-并列结构
    choice = random.choices([0, 1], weights=[2, 1], k=1)[0]  # 增加嵌套结构的概率
    
    if choice == 0:
        # 生成嵌套结构
        left, right = random.choice([('(', ')'), ('[', ']'), ('<', '>')])
        inner = generate_brackets(max_depth - 1)
        return left + inner + right
    else:
        # 生成并列结构，随机生成2-3个并列部分
        parts = [generate_brackets(max_depth - 1) for _ in range(random.randint(2, 3))]
        random.shuffle(parts)  # 打乱顺序增加复杂度
        return ''.join(parts)

def is_valid(s):
    """
    验证括号字符串是否合法
    :param s: 输入的字符串
    :return: 是否合法
    """
    stack = []
    pairs = {')': '(', ']': '[', '>': '<'}
    for char in s:
        if char in '([<':
            stack.append(char)
        else:
            if not stack or pairs[char] != stack.pop():
                return False
    return not stack

# 生成10个示例并验证
parentheses = [generate_brackets(max_depth=3) for _ in range(2000)]
```
