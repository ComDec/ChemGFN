```python
import pandas as pd
```

```python
df = pd.read_csv("/data1/xw3763/project/gflow/ChemGFN/notebooks/all_TrainTestVal_pXC50.csv")
```

```python
df.head()
```

```python
# extract smiles column
smiles = df["SMILES"].tolist()
smiles = [x.split("|")[-1] for x in smiles]
```

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-1B")
```

```python
all_tokens = []
for smi in smiles:
    tokens = tokenizer.encode(smi, add_special_tokens=False)
    all_tokens.extend(tokens)

all_tokens = list(set(all_tokens))
```

```python
for token in all_tokens:
    print(tokenizer.decode([token]))
    
with open("/data1/xw3763/project/gflow/ChemGFN/assets/token_list/SMILES/allowed_llama3.2_1B_allowed_token_v1", "w") as f:
    for token in all_tokens:
        f.write(tokenizer.decode([token]) + "\n")
```
