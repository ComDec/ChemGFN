```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import torch

# 假设输入的两个向量
data1 = torch.load("/home/wangxi/project/LLM/ChemGFN/before.pt")
data2 = torch.load("/home/wangxi/project/LLM/ChemGFN/after.pt")
```

```python
data2.shape
```

```python
torch.isinf(data2[0]).sum()
```

```python
# 创建数据集并添加标识
import pandas as pd
df = pd.DataFrame({
    'value': np.concatenate([data2[0], data2[0]]),
    'group': ['Before'] * data1.shape[-1] + ['After'] * data2.shape[-1]
})

# 使用 seaborn 的 displot 绘制概率密度图
sns.displot(df, x='value', hue='group', kind='kde', fill=True, palette='Set1')

# 设置标题和标签
plt.title('Density')
plt.xlabel('Token Ids')
plt.ylabel('Logits')

# 显示图像
plt.show()
```
