```python
import json
import csv, os
json_path = "/home/wangxi/project/LLM/ChemGFN/runs/fix_cfg/wandb/latest-run/files/media/table/samples"

file_list = os.listdir(json_path)

for idx, file in enumerate(file_list):
    json_file= os.path.join(json_path, file)
    
    file_name = file.split('.')[0]
    # 读取json文件
    with open(json_file, 'r') as f:
        data = json.load(f)

    # 提取列名
    columns = data['columns']

    # 提取数据
    data_rows = data['data']

    # 写入csv文件
    with open(os.path.join(json_path, f'{file_name}.csv'), 'w', newline='') as csv_file:
        writer = csv.writer(csv_file)
        writer.writerow(columns)
        for row in data_rows:
            writer.writerow(row)
```
