```python
import numpy as np
meta_data = np.load("dataset.npy")
```

```python
train_data = meta_data[:int(len(meta_data) * 0.95)]
val_data = meta_data[int(len(meta_data) * 0.95):]
```

```python
[str(x) for x in train_data[0]]
```

```python
train_dict = []
val_dict = []
for i in range(len(train_data)):
    train_dict.append({
        "prompt": "Generate a list of numbers that follow the rule: even number followed by odd number and odd number followed by even number and between 0 and 20: ",
        "output": [str(x) for x in train_data[i]]
    })
    
for i in range(len(val_data)):
    val_dict.append({
        "prompt": "Generate a list of numbers that follow the rule: even number followed by odd number and odd number followed by even number and between 0 and 20: ",
        "output": [str(x) for x in val_data[i]]
    })
```

```python
train_dict[0]
```

```python
from datasets import Dataset
train_dataset = Dataset.from_list(train_dict)
val_dataset = Dataset.from_list(val_dict)
```

```python
train_dataset.to_json("train_data.json")
val_dataset.to_json("val_data.json")
```

```python
train_dataset.push_to_hub("EscheWang/AlternateNumbers", split="train")
val_dataset.push_to_hub("EscheWang/AlternateNumbers", split="val")
```
